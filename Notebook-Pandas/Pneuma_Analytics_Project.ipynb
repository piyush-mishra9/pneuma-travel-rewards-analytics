# Monthly Search Trend 📈

```python
# Convert search_date to datetime format
searches["search_date"] = pd.to_datetime(searches["search_date"])
```

```python
# Analyze the number of searches by month
searches["search_date"].dt.to_period("M").value_counts().sort_index()
```

# Booking Conversion Rate

```python
# Calculate the search-to-booking conversion rate
conversion_rate = (bookings["search_id"].nunique() / searches["search_id"].nunique()) * 100

conversion_rate

# Conversion Rate = Unique searches that resulted in a booking ÷ Total unique searches × 100
```

# Subscription Analysis 💰

```python
# Analyze the distribution of subscription plans
subscriptions["plan"].value_counts()
```

```python
# Analyze revenue generated by each subscription plan
subscriptions.groupby("plan")["revenue_inr"].sum().sort_values(ascending=False)
```

```python
# Merge subscription data with user information
user_subscription = subscriptions.merge(
    users[["user_id", "user_type"]],
    on="user_id",
    how="left"
)
```

```python
# Analyze subscriptions by user type
user_subscription["user_type"].value_counts()
```

```python
# Calculate the subscription rate for each user type
subscription_rate = (
    user_subscription.groupby("user_type")["user_id"]
    .nunique()
    .div(users.groupby("user_type")["user_id"].nunique())
    * 100
)

subscription_rate.sort_values(ascending=False)
```

```python
# Analyze total miles redeemed by each airline
bookings.merge(
    award_results[["result_id", "airline"]],
    on="result_id",
    how="left"
).groupby("airline")["miles_used"].sum().sort_values(ascending=False)
```

```python
# Calculate the total miles redeemed
bookings["miles_used"].sum()
```

```python
# Calculate the total taxes paid for bookings
bookings["taxes_paid_inr"].sum()
```

```python
# Calculate the average miles redeemed per booking
bookings["miles_used"].mean()
```

```python
# Calculate the average taxes paid per booking
bookings["taxes_paid_inr"].mean()
```

```python
# Identify users who performed more than one search
searches_per_user = searches.groupby("user_id")["search_id"].count()

repeat_users = (searches_per_user > 1).sum()

repeat_users
```

```python
# Identify the most active users based on search activity
searches_per_user.sort_values(ascending=False).head(10)
```

```python
# Analyze the routes with the highest number of bookings
bookings.merge(
    searches[["search_id", "origin", "destination"]],
    on="search_id",
    how="left"
).groupby(
    ["origin", "destination"]
).size().sort_values(ascending=False).head(10)
```
